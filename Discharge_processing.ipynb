# Preliminaries
This jupyter notebook documents the data processing workflow for the UAWOS discharge calculation. We demonstrate the data processing workflow for one example from the Po River. We only include basic discharge calculation steps. All necessary input data, and code files to run this notebook are included in a folder "Discharge_processing_example", which is distributed with this notebook and is expected to sit in the directory from which this notebook is run. The Discharge_processing_example folder also contains a listing of the UAWOS python environment and a corresponding .yaml file. Please install this environment (e.g. using conda) prior to running the notebook. 

# Input Data Description
The discharge calculation requires three input datasets.
Below, each input file is described using the filenames provided in the example folder:

* {XSpo2}.csv: WPR Level-3 data from the UAWOS project.

* {XSpo2}_forQ.csv: Doppler Level-3 data from the UAWOS project.

* {XSpo2}_processed.csv: WPR Level-4 data from the UAWOS project.

Additionally, a subfolder named logo is included in the example directory.
This folder is **optional** and used only for embedding UAWOS logo in the automatically generated discharge report.

# Processing Steps

## 1. Necessary modules import
The following cell loads all the required packages and modules.

```python
import os
import sys
sys.path.append(os.getcwd())
from UAWOS_Discharge import UAWOS_Discharge
```

## 2. General attributies setup


| No. | Parameter                   | Description                                                                                                                                                                                                                                                                                                                                                                                                                                                                                          | Unit / Example                                         |
| --- | --------------------------- | ---------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------- | ------------------------------------------------------ |
| 1   | **project_path**          | Folder path where all input and output files will be saved.                                                                                                                                                                                                                                                                                                                                                                                                                                          | e.g., `C:\Discharge_processing_example\Discharge_processing_example`                    |
| 2   | **XSid**                  | Identifier of the calculated cross section. It will appear in the report title and output filenames.                                                                                                                                                                                                                                                                                                                                                                                            | e.g., `Po2`                                           |
| 3   | **xy_EPSG**, **z_EPSG** | EPSG codes for coordinate reference systems (CRS).                                                                                                                                                                                                                                                                                                                                                                                                                  | e.g., `32632` (UTM Zone 32N), `7837` (DHHN2016 height) |
| 4   | **dx**             | The width of each water column (bin) used in the mid-section discharge calculation method. A smaller `dx` yields finer spatial resolution but higher computational cost. For detailed methodology, please refer to [**here**](https://essopenarchive.org/users/966819/articles/1335195-contactless-river-discharge-from-unoccupied-aerial-systems-hydrometry-performance-evaluation-using-a-large-and-diverse-set-of-river-cross-sections). |                                *m*                        | 


```python
general_attributes = dict()
general_attributes['project_path'] = os.getcwd() + os.sep + 'Discharge_processing_example'
general_attributes['XSid'] = 'Po2' # index of calculated cross section
general_attributes['xy_EPSG'] = 'EPSG:3006'
general_attributes['z_EPSG'] = 'EPSG:5613'
general_attributes['dx'] = 1 # Width of water coulmn
```

## 3. Calculation attributions setup
These parameters define the input data sources and computation options used in discharge estimation and uncertainty analysis.
| No. | Parameter                 | Description                                                                           | Unit / Example                                      |
| --- | ------------------------- | ------------------------------------------------------------------------------------- | --------------------------------------------------- |
| 1   | **velocity**            | Path to Doppler Level-3 input data.                                                   | e.g., `D:\UAWOS\Dataset\Velocity\XSpo2_forQ.csv.csv`   |
| 2   | **bathymetry**          | Path to WPR Level-3 bathymetry input data.                                            | e.g., `D:\UAWOS\Dataset\Bathymetry\XSpo2.csv.csv` |
| 3   | **full_geometry**       | Path to WPR Level-4 full-geometry input data.                                         | e.g., `D:\UAWOS\Dataset\Geometry\XSpo2_processed.csv`   |
| 4   | **WSE**                 | Water Surface Elevation.                                                              | *m*                                                 |
| 5   | **WSS**                 | Water Surface Slope.                                                                  | *cm / km*                                           |
| 6   | **WSE_sigma**           | Standard deviation of WSE.                                                            | *m*                                                 |
| 7   | **WSS_sigma**           | Standard deviation of WSS.                                                            | *cm / km*                                           |
| 8   | **lbguess**             | Rough estimate of the left land-water boundary along the cross-sectional coordinate.  | *m*                                                 |
| 9   | **rbguess**             | Rough estimate of the right land-water boundary along the cross-sectional coordinate. | *m*                                                 |
| 10  | **if_calc_uncertainty** | Whether to perform uncertainty calculation.                            | `True` / `False`                                    |
| 11  | **sample_size**         | Number of Monte Carlo samples used if uncertainty analysis is enabled.                | e.g., `1000`                                        |


```python
cal_attributes = general_attributes.copy()
cal_attributes['resultpath'] = general_attributes['project_path'] + os.sep + 'result'
cal_attributes['velocity'] = general_attributes['project_path'] + os.sep + 'XSpo2_forQ.csv'
cal_attributes['bathymetry'] = general_attributes['project_path'] + os.sep + 'XSpo2.csv'
cal_attributes['full_geometry'] =general_attributes['project_path'] + os.sep + 'XSpo2_processed.csv'
cal_attributes['WSE'] = 179.1131     # Water surface elevation (m)
cal_attributes['WSS'] = 129.3703     # Water surface slope (cm/km)
cal_attributes['WSE_sigma'] = 0.03   # Standard deviation of WSE
cal_attributes['WSS_sigma'] = 0.4264 # Standard deviation of WSS
cal_attributes['lbguess'] = -58      # Rough estimate of the land-water left boundary
cal_attributes['rbguess'] = 47       # Rough estimate of the land-water right boundary
cal_attributes['if_calc_uncertainty'] = True # Whether calculate the uncertainty
                                             # related the observation data, if true.
cal_attributes['sample_size'] = 3 # sample size of Monte Carlo simulation
```

## 4. Discharge calculation

```python
XS = UAWOS_Discharge(cal_attributes)
result = XS.process_Discharge()
```
